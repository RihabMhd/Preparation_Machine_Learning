```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.DataFrame({
    'Age': [22, 25, 30, 35, 40, 28, 45, 32, 26, 38],
    'Salaire' : [3000, 4500, 6000, 7500, 9000, 5000, 10000, 6500, 4000, 8500],
    'Ville' : ['Agadir', 'Casa', 'Rabat', 'Agadir', 'Casa',
              'Rabat', 'Agadir', 'Casa', 'Rabat', 'Agadir'],
    'A_achete' : [0, 0, 1, 1, 1, 0, 1, 1, 0, 1]
})
df
```

```python
y = df["A_achete"]
x = df.drop("A_achete",axis=1)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x,
    y,
    random_state=42,
    test_size=0.2,
    stratify=y
)
```

```python
columnsNumerique = x.select_dtypes(include=["number"]).columns.tolist()
columnsCategories = x.select_dtypes(include=["object"]).columns.tolist()
```

```python
pre = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), columnsNumerique),
        ("cat", OneHotEncoder(), columnsCategories)
    ]
)
```

```python
pipeline = Pipeline([
    ("preprocessing", pre),
    ("model", LogisticRegression())
])

pipeline.fit(X_train, y_train)
y_pred = pipeline.predict(X_test)

acc = accuracy_score(y_test, y_pred)
print(acc)
```
